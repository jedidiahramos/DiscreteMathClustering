```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
from google.colab import files

# List of clustering algorithm columns
clustering_columns = [
    'cluster_kmeans', 'cluster_dbscan', 'cluster_optics', 'cluster_hdbscan',
    'cluster_spectral', 'cluster_gmm', 'cluster_agglo', 'cluster_meanshift',
    'cluster_affinity', 'cluster_birch', 'h_cluster'
]

# Evaluation metrics to compare
evaluation_metrics = [
    'silhouette_score', 'average_purity',
    'weighted_average_purity', 'ari', 'nmi', 'v_score'
]


def extract_summary_table(df):
    # Find index where 'cluster' == 'birch'
    end_idx = df[df['cluster'] == 'birch'].index.max()

    # Include all rows up to and including 'birch'
    return df.loc[:end_idx].copy()


def insert_empty_rows_between_clusters(df, cluster_col='Cluster'):
    df_with_empty_rows = []

    # Get unique cluster values in order
    clusters = df[cluster_col].unique()

    for cluster in clusters:
        cluster_rows = df[df[cluster_col] == cluster]
        df_with_empty_rows.append(cluster_rows)

        # Add an empty row (all columns blank)
        empty_row = pd.DataFrame([{col: "" for col in df.columns}])
        df_with_empty_rows.append(empty_row)

    # Concatenate all pieces together
    return pd.concat(df_with_empty_rows, ignore_index=True)


# Function to plot a table
def plot_table(df, title='Comparison Table'):
    fig, ax = plt.subplots(figsize=(18, 0.6 * len(df)))
    ax.axis('off')
    table = ax.table(
        cellText=df.values,
        colLabels=df.columns,
        loc='center',
        cellLoc='center'
    )
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)
    plt.title(title, fontsize=14, pad=10)
    plt.subplots_adjust(top=0.9, bottom=0.05)
    plt.show()
```

#Learning Outcomes (Euclidean vs Cosine)

```python
# Upload two CSVs: Euclidean first, then Cosine ... or Unprocessed first, then Processed
uploaded = files.upload()
file1, file2 = list(uploaded.keys())

# Load and tag type
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)
df1['sheet_type'] = 'Euclidean'
df2['sheet_type'] = 'Cosine'

# Combine for plotting
cols = ['Category', 'Index', 'x', 'y', 'sheet_type'] + clustering_columns
df1 = df1[cols]
df2 = df2[cols]
df = pd.concat([df1, df2], ignore_index=True)

# Add hover label
df['label'] = "#" + df['Index'].astype(str) + " in " + df['Category']

# Plot with Plotly
fig = px.scatter(
    df, x='x', y='y', color='sheet_type',
    hover_name='label',
    hover_data={'Category': True, 'Index': True, 'x': True, 'y': True},
    title='2D Embeddings: Euclidean vs Cosine'
)
fig.update_traces(marker=dict(size=8, opacity=0.6))
fig.show()

# Combine both datasets for plotting
df_plot = pd.concat([df1, df2], ignore_index=True)

# Step 1: Plot each algorithm comparison
for col in clustering_columns:
    plt.figure(figsize=(10, 6))

    # Plot Sheet 1 values
    sheet1_values = df1[col]
    plt.scatter(df1.index, sheet1_values, color='blue', label='Euclidean', alpha=0.6)

    # Plot Sheet 2 values
    sheet2_values = df2[col]
    plt.scatter(df2.index, sheet2_values, color='red', label='Cosine', alpha=0.6)

    # Customize plot
    plt.title(f'Comparison of {col} between Euclidean and Cosine')
    plt.xlabel('Row Index')
    plt.ylabel(f'{col} Values')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Load data again
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

df1_summary = extract_summary_table(df1)
df2_summary = extract_summary_table(df2)

# Proceed with merge and metric comparison
df_merged = pd.merge(
    df1_summary, df2_summary,
    on='cluster',
    suffixes=('_eu', '_cos')
)

# Collect comparison data
comparison_data = []
for _, row in df_merged.iterrows():
    cluster = row['cluster']
    for evaluation_metric in evaluation_metrics:
        val_eu = row.get(f'{evaluation_metric}_eu', None)
        val_cos = row.get(f'{evaluation_metric}_cos', None)

        try:
            if pd.isna(val_eu) or pd.isna(val_cos):
                abs_diff = None
                winner = 'N/A'
            else:
                abs_diff = abs(val_eu - val_cos)
                winner = (
                    'Euclidean' if val_eu > val_cos
                    else 'Cosine' if val_cos > val_eu
                    else 'Tie'
                )
        except:
            abs_diff = None
            winner = 'N/A'

        comparison_data.append([
            cluster, evaluation_metric, val_eu, val_cos, abs_diff, winner
        ])

# Create DataFrame
df_comp = pd.DataFrame(comparison_data, columns=[
    'Cluster', 'Evaluation Metric', 'Euclidean', 'Cosine', 'Absolute Difference', 'Winner'
])

# Show table as a plot
df_comp_spaced = insert_empty_rows_between_clusters(df_comp)
plot_table(df_comp_spaced, title="Comparison Table by Cluster between Euclidean and Cosine for Learning Outcomes")
```

#Others

```python
def compare_cluster_data(is_euclidean_vs_cosine, sheet_name):
  # Upload two CSVs: Euclidean first, then Cosine ... or Unprocessed first, then Processed
  uploaded = files.upload()
  file1, file2 = list(uploaded.keys())

  # Load data
  df1 = pd.read_csv(file1)
  df2 = pd.read_csv(file2)

  sheet_type1 = 'Unprocessed'
  sheet_type2 = 'Processed'
  if is_euclidean_vs_cosine:
    sheet_type1 = 'Euclidean'
    sheet_type2 = 'Cosine'

  df1['sheet_type'] = sheet_type1
  df2['sheet_type'] = sheet_type2

  # Combine for plotting
  cols = ['category', 'cleaned_question', 'x', 'y', 'sheet_type'] + clustering_columns
  df1 = df1[cols]
  df2 = df2[cols]
  df = pd.concat([df1, df2], ignore_index=True)

  # Add hover label
  df['label'] = df['category'] + ": " + df['cleaned_question']

  # Plot with Plotly
  fig = px.scatter(
      df, x='x', y='y', color='sheet_type',
      hover_name='label',
      hover_data={'category': True, 'cleaned_question': True, 'x': True, 'y': True},
      title='2D Embeddings: ' + sheet_type1 + ' vs ' + sheet_type2 + ' for ' + sheet_name
  )
  fig.update_traces(marker=dict(size=8, opacity=0.6))
  fig.show()

  # Combine both datasets for plotting
  df_plot = pd.concat([df1, df2], ignore_index=True)

  # Step 1: Plot each algorithm comparison
  for col in clustering_columns:
      plt.figure(figsize=(10, 6))

      # Plot Sheet 1 values
      sheet1_values = df1[col]
      plt.scatter(df1.index, sheet1_values, color='blue', label=sheet_type1, alpha=0.6)

      # Plot Sheet 2 values
      sheet2_values = df2[col]
      plt.scatter(df2.index, sheet2_values, color='red', label=sheet_type2, alpha=0.6)

      # Customize plot
      plt.title(f'Comparison of {col} between {sheet_type1} and {sheet_type2} for {sheet_name}')
      plt.xlabel('Row Index')
      plt.ylabel(f'{col} Values')
      plt.legend()
      plt.grid(True)
      plt.tight_layout()
      plt.show()


def compare_evaluation_metrics(is_euclidean_vs_cosine, sheet_name):
  # Upload two CSVs: Euclidean first, then Cosine ... or Unprocessed first, then Processed
  uploaded = files.upload()
  file1, file2 = list(uploaded.keys())

  # Load data
  df1 = pd.read_csv(file1)
  df2 = pd.read_csv(file2)

  df1_summary = extract_summary_table(df1)
  df2_summary = extract_summary_table(df2)

  sheet_type1 = 'Unprocessed'
  sheet_type2 = 'Processed'
  if is_euclidean_vs_cosine:
    sheet_type1 = 'Euclidean'
    sheet_type2 = 'Cosine'

  # Proceed with merge and sheet comparison
  df_merged = pd.merge(
      df1_summary, df2_summary,
      on='cluster',
      suffixes=('_' + sheet_type1, '_' + sheet_type2)
  )

  # Collect comparison data
  comparison_data = []
  for _, row in df_merged.iterrows():
      cluster = row['cluster']
      for evaluation_metric in evaluation_metrics:
          val1 = row.get(f'{evaluation_metric}_' + sheet_type1, None)
          val2 = row.get(f'{evaluation_metric}_' + sheet_type2, None)

          try:
              if pd.isna(val1) or pd.isna(val2):
                  abs_diff = None
                  winner = 'N/A'
              else:
                  abs_diff = abs(val1 - val2)
                  winner = (
                      sheet_type1 if val1 > val2
                      else sheet_type2 if val2 > val1
                      else 'Tie'
                  )
          except:
              abs_diff = None
              winner = 'N/A'

          comparison_data.append([
              cluster, evaluation_metric, val1, val2, abs_diff, winner
          ])

  # Create DataFrame
  df_comp = pd.DataFrame(comparison_data, columns=[
      'Cluster', 'Evaluation Metric', sheet_type1, sheet_type2, 'Absolute Difference', 'Winner'
  ])

  # Show table as a plot
  df_comp_spaced = insert_empty_rows_between_clusters(df_comp)
  plot_table(df_comp_spaced, title="Comparison Table by Cluster between " + sheet_type1 + " and " + sheet_type2 + " for " + sheet_name)
```
